```python
import pandas as pd
import numpy as np

df = pd.read_csv("iris.csv")
df
```

```python
print(df.shape)
print(df.columns)
print(df.head())
```

```python
df.describe()
```

```python
print("Mean:")
print(df["sepal.length"].mean())
print(df["sepal.width"].mean())
print(df["petal.length"].mean())
print(df["petal.width"].mean())
```

```python
print("Standard Deviation:")
print(df["sepal.length"].std())
print(df["sepal.width"].std())
print(df["petal.length"].std())
print(df["petal.width"].std())
```

```python
print("Covariance:")
print("1 & 2:", df["sepal.length"].cov(df["sepal.width"]))
print("1 & 3:", df["sepal.length"].cov(df["petal.length"]))
print("1 & 4:", df["sepal.length"].cov(df["petal.width"]))
```

```python
print("Correlation:")
print("1 & 2:", df["sepal.length"].corr(df["sepal.width"]))
print("1 & 3:", df["sepal.length"].corr(df["petal.length"]))
print("1 & 4:", df["sepal.length"].corr(df["petal.width"]))
```
