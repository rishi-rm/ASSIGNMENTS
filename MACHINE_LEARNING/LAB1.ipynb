```python
np.random.seed(42)

arr = np.random.randint(1, 101, size=(10, 10))

print(arr)
print(arr.shape)
```

```python
A = arr[:5, :7]
B = arr[:7, 5:]
C = arr[5:, :7]

print("A:")
print(A)
print("Shape of A:", A.shape)

print("\nB:")
print(B)
print("Shape of B:", B.shape)

print("\nC:")
print(C)
print("Shape of C:", C.shape)
```

```python
sum_AC = A + C

print("A + C:")
print(sum_AC)
```

```python
multiplication_AB = A @ B

print("A × B:")
print(multiplication_AB)
```

```python
np.random.seed(100)

arr2 = np.random.randint(1, 101, size=(10, 10))

print(arr2)
print(arr2.shape)
```
