```python
import seaborn as sns
```

```python
dataset = sns.load_dataset('titanic')
```

```python
dataset
```

```python
dataset.info()
```

```python
sns.histplot(data=dataset, x="age", hue="survived", multiple="stack")
```

```python
sns.scatterplot(data=dataset, x="age", y="fare", hue="survived")
```

```python
sns.histplot(data=dataset, x="age", hue="survived", shrink=.8, alpha=.8, legend=False)
```

```python
import matplotlib.pyplot as plt
```

```python
sns.scatterplot(data=dataset, x="age", y="fare", hue="sex", style="survived")
plt.title("Age vs Fare")
plt.show()
```

```python
sns.scatterplot(data=dataset, x="age", y="fare", hue="sex", size="pclass")
plt.title("Age vs Fare")
plt.show()
```

```python
sns.lineplot(
    data=dataset,
    x="age",
    y="fare"
)
plt.title("Age vs Fare - LINE PLOT")
plt.show()
```

```python
sns.relplot(
    data=dataset,
    x="age",
    y="fare",
    hue="sex",
    col="class",
    kind="scatter"
)

plt.show()
```

# Q2. flights dataset

```python
flights_dataset = sns.load_dataset('flights')
```

```python
flights_dataset
```

```python
sns.scatterplot(
    flights_dataset,
    x="year",
    y = "passengers"
)
plt.title("scatterplot of year to passengers")
plt.show()
```

```python
sns.scatterplot(
    flights_dataset,
    x="year",
    y = "passengers",
    hue='month'
)
plt.title("scatterplot of year to passengers")
plt.show()
```

```python
sns.lineplot(
    data=flights_dataset,
    x="year",
    y="passengers",
    hue='month'
    
)

plt.title("Passengers Over Years")
plt.show()
```

## grids

```python
sns.pairplot(flights_dataset)
plt.show()
```

```python
flights_dataset["month_num"] = flights_dataset["month"].map({
    "Jan": 1,
    "Feb": 2,
    "Mar": 3,
    "Apr": 4,
    "May": 5,
    "Jun": 6,
    "Jul": 7,
    "Aug": 8,
    "Sep": 9,
    "Oct": 10,
    "Nov": 11,
    "Dec": 12
})
```

```python
sns.pairplot(flights_dataset, vars=['year', 'month_num', 'passengers']) 
plt.show()
```

```python
g = sns.PairGrid(
    flights_dataset,
    vars=["year", "month_num", "passengers"]
)

g.map(sns.scatterplot)

plt.show()
```

```python
g = sns.PairGrid(
    flights_dataset,
    vars=["year", "month_num", "passengers"]
)

g.map_upper(sns.scatterplot)
g.map_diag(sns.histplot)
g.map_lower(sns.lineplot)

plt.show()
```
